In industry, data science teams frequently build multiple models to address different situations within the same project. This concept is called Model Segmentation. We will use this exact concept to resolve the extreme errors we discovered in our premium prediction project.

The idea behind model segmentation is to divide our data based on the problematic segments and train specialized models for each.

<h3>Importing required libraries</h3>

```python
import pandas as pd
```

<h3>Load and segment the data</h3>

```python
df = pd.read_excel('data/premiums.xlsx')
```

Segmenting by a 25-year age threshold involves splitting your dataset into two distinct groups: individuals under 25 and those 25 and older. In pandas, you can easily achieve this using boolean indexing to create two separate DataFrames.

```python
df_under_25 = df[df['Age'] < 25]
df_over_25 = df[df['Age'] >= 25]
```

<h3>Save the data into two files</h3>

```python
df_under_25.to_excel('data/premiums young.xlsx', index = False)
df_over_25.to_excel('data/premiums adults.xlsx', index = False)
```
